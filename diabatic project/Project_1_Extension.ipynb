```python
# ============================================================
# PHASE 4 — FEATURE ENGINEERING
# ============================================================

import numpy as np
import pandas as pd

# Use KNN-imputed data as our improved dataset
df_fe = X_knn.copy()

# Add target
df_fe["Outcome"] = y_extension.values

# ------------------------------------------------------------
# 1. BMI Category
# ------------------------------------------------------------
df_fe["BMI_Category"] = pd.cut(
    df_fe["BMI"],
    bins=[-np.inf, 18.5, 25, 30, np.inf],
    labels=[0, 1, 2, 3]
).astype(int)

# ------------------------------------------------------------
# 2. Age Group
# ------------------------------------------------------------
df_fe["Age_Group"] = pd.cut(
    df_fe["Age"],
    bins=[-np.inf, 30, 50, np.inf],
    labels=[0, 1, 2]
).astype(int)

# ------------------------------------------------------------
# 3. Glucose Category
# ------------------------------------------------------------
df_fe["Glucose_Category"] = pd.cut(
    df_fe["Glucose"],
    bins=[-np.inf, 99, 125, np.inf],
    labels=[0, 1, 2]
).astype(int)

# ------------------------------------------------------------
# 4. Interaction Features
# ------------------------------------------------------------
df_fe["Glucose_BMI"] = df_fe["Glucose"] * df_fe["BMI"]

df_fe["Glucose_Age"] = df_fe["Glucose"] * df_fe["Age"]

df_fe["BMI_Age"] = df_fe["BMI"] * df_fe["Age"]

# ------------------------------------------------------------
# 5. Ratio Feature
# ------------------------------------------------------------
df_fe["Glucose_BMI_Ratio"] = (
    df_fe["Glucose"] / (df_fe["BMI"] + 1e-6)
)

# ------------------------------------------------------------
# 6. Metabolic Risk Index
# ------------------------------------------------------------
df_fe["Metabolic_Risk_Index"] = (
    (df_fe["Glucose"] >= 126).astype(int)
    + (df_fe["BMI"] >= 30).astype(int)
    + (df_fe["BloodPressure"] >= 80).astype(int)
    + (df_fe["Age"] >= 45).astype(int)
)

# ------------------------------------------------------------
# Display results
# ------------------------------------------------------------

print("Feature Engineering Completed!")

print("\nOriginal number of features:", X_knn.shape[1])

print("New number of features:", df_fe.drop(
    "Outcome", axis=1
).shape[1])

print("\nNew Features:")
new_features = [
    "BMI_Category",
    "Age_Group",
    "Glucose_Category",
    "Glucose_BMI",
    "Glucose_Age",
    "BMI_Age",
    "Glucose_BMI_Ratio",
    "Metabolic_Risk_Index"
]

print(new_features)

print("\nDataset after Feature Engineering:")
display(df_fe.head())
```

```python
# ============================================================
# PHASE 5 — FEATURE SELECTION
# ============================================================

from sklearn.feature_selection import mutual_info_classif
from sklearn.ensemble import RandomForestClassifier

# Separate features and target
X_fe = df_fe.drop("Outcome", axis=1)
y_fe = df_fe["Outcome"]

# ------------------------------------------------------------
# 1. Mutual Information
# ------------------------------------------------------------

mi_scores = mutual_info_classif(
    X_fe,
    y_fe,
    random_state=42
)

mi_df = pd.DataFrame({
    "Feature": X_fe.columns,
    "Mutual_Information": mi_scores
})

mi_df = mi_df.sort_values(
    "Mutual_Information",
    ascending=False
)

print("Mutual Information Scores:")
display(mi_df)

# ------------------------------------------------------------
# 2. Random Forest Feature Importance
# ------------------------------------------------------------

rf_selector = RandomForestClassifier(
    n_estimators=200,
    random_state=42
)

rf_selector.fit(X_fe, y_fe)

importance_df = pd.DataFrame({
    "Feature": X_fe.columns,
    "Importance": rf_selector.feature_importances_
})

importance_df = importance_df.sort_values(
    "Importance",
    ascending=False
)

print("\nRandom Forest Feature Importance:")
display(importance_df)

# ------------------------------------------------------------
# Plot feature importance
# ------------------------------------------------------------

plt.figure(figsize=(10, 7))

sns.barplot(
    data=importance_df,
    x="Importance",
    y="Feature"
)

plt.title("Feature Importance")
plt.xlabel("Importance")
plt.ylabel("Feature")

plt.show()
```

```python
# ============================================================
# PHASE 6 — CLASS IMBALANCE HANDLING
# ============================================================

from sklearn.model_selection import train_test_split
from imblearn.over_sampling import SMOTE

# Use engineered features
X = df_fe.drop("Outcome", axis=1)
y = df_fe["Outcome"]

# ------------------------------------------------------------
# Train/Test Split
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training set:", X_train.shape)
print("Testing set:", X_test.shape)

# ------------------------------------------------------------
# Original class distribution
# ------------------------------------------------------------

print("\nOriginal training class distribution:")
print(y_train.value_counts())

# ------------------------------------------------------------
# SMOTE
# ------------------------------------------------------------

smote = SMOTE(
    random_state=42
)

X_train_smote, y_train_smote = smote.fit_resample(
    X_train,
    y_train
)

print("\nAfter SMOTE:")
print(y_train_smote.value_counts())

print("\nBefore SMOTE:", X_train.shape)
print("After SMOTE:", X_train_smote.shape)
```

```python
# ============================================================
# PHASE 7 — HYPERPARAMETER TUNING
# ============================================================

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.model_selection import RandomizedSearchCV, StratifiedKFold

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# ------------------------------------------------------------
# 1. Logistic Regression
# ------------------------------------------------------------

logistic_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(max_iter=2000))
])

logistic_params = {
    "model__C": [0.01, 0.1, 1, 10, 100],
    "model__class_weight": [None, "balanced"]
}

logistic_search = RandomizedSearchCV(
    logistic_pipeline,
    logistic_params,
    n_iter=10,
    scoring="f1",
    cv=cv,
    random_state=42,
    n_jobs=-1
)

logistic_search.fit(X_train_smote, y_train_smote)

# ------------------------------------------------------------
# 2. KNN
# ------------------------------------------------------------

knn_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", KNeighborsClassifier())
])

knn_params = {
    "model__n_neighbors": list(range(3, 16)),
    "model__weights": ["uniform", "distance"],
    "model__p": [1, 2]
}

knn_search = RandomizedSearchCV(
    knn_pipeline,
    knn_params,
    n_iter=10,
    scoring="f1",
    cv=cv,
    random_state=42,
    n_jobs=-1
)

knn_search.fit(X_train_smote, y_train_smote)

# ------------------------------------------------------------
# 3. Decision Tree
# ------------------------------------------------------------

dt = DecisionTreeClassifier(
    random_state=42
)

dt_params = {
    "max_depth": [3, 5, 7, 10, None],
    "min_samples_split": [2, 5, 10, 20],
    "min_samples_leaf": [1, 2, 5, 10],
    "class_weight": [None, "balanced"]
}

dt_search = RandomizedSearchCV(
    dt,
    dt_params,
    n_iter=15,
    scoring="f1",
    cv=cv,
    random_state=42,
    n_jobs=-1
)

dt_search.fit(X_train_smote, y_train_smote)

# ------------------------------------------------------------
# 4. Random Forest
# ------------------------------------------------------------

rf = RandomForestClassifier(
    random_state=42
)

rf_params = {
    "n_estimators": [100, 200, 300, 500],
    "max_depth": [None, 5, 10, 15, 20],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 4],
    "max_features": ["sqrt", "log2"],
    "class_weight": [None, "balanced"]
}

rf_search = RandomizedSearchCV(
    rf,
    rf_params,
    n_iter=20,
    scoring="f1",
    cv=cv,
    random_state=42,
    n_jobs=-1
)

rf_search.fit(X_train_smote, y_train_smote)

# ------------------------------------------------------------
# 5. SVM
# ------------------------------------------------------------

svm_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", SVC(probability=True))
])

svm_params = {
    "model__C": [0.01, 0.1, 1, 10, 100],
    "model__gamma": ["scale", "auto", 0.001, 0.01, 0.1],
    "model__kernel": ["linear", "rbf", "poly"],
    "model__class_weight": [None, "balanced"]
}

svm_search = RandomizedSearchCV(
    svm_pipeline,
    svm_params,
    n_iter=20,
    scoring="f1",
    cv=cv,
    random_state=42,
    n_jobs=-1
)

svm_search.fit(X_train_smote, y_train_smote)

# ------------------------------------------------------------
# Best Parameters
# ------------------------------------------------------------

search_results = {
    "Logistic Regression": logistic_search,
    "KNN": knn_search,
    "Decision Tree": dt_search,
    "Random Forest": rf_search,
    "SVM": svm_search
}

for name, search in search_results.items():
    print("\n", name)
    print("Best Parameters:")
    print(search.best_params_)
    print("Best CV F1:", search.best_score_)
```

```python
# ============================================================
# PHASE 8 — STRATIFIED CROSS-VALIDATION
# ============================================================

from sklearn.model_selection import cross_val_score

tuned_models = {
    "Logistic Regression": logistic_search.best_estimator_,
    "KNN": knn_search.best_estimator_,
    "Decision Tree": dt_search.best_estimator_,
    "Random Forest": rf_search.best_estimator_,
    "SVM": svm_search.best_estimator_
}

cv_results = []

for name, model in tuned_models.items():

    scores = cross_val_score(
        model,
        X_train_smote,
        y_train_smote,
        cv=cv,
        scoring="f1"
    )

    cv_results.append({
        "Model": name,
        "Mean CV F1": scores.mean(),
        "Std CV F1": scores.std()
    })

cv_results_df = pd.DataFrame(cv_results)

cv_results_df = cv_results_df.sort_values(
    "Mean CV F1",
    ascending=False
)

print("Cross-Validation Results:")
display(cv_results_df)
```

```python
# ============================================================
# PHASE 8 — STRATIFIED CROSS-VALIDATION
# ============================================================

from sklearn.model_selection import cross_val_score

tuned_models = {
    "Logistic Regression": logistic_search.best_estimator_,
    "KNN": knn_search.best_estimator_,
    "Decision Tree": dt_search.best_estimator_,
    "Random Forest": rf_search.best_estimator_,
    "SVM": svm_search.best_estimator_
}

cv_results = []

for name, model in tuned_models.items():

    scores = cross_val_score(
        model,
        X_train_smote,
        y_train_smote,
        cv=cv,
        scoring="f1"
    )

    cv_results.append({
        "Model": name,
        "Mean CV F1": scores.mean(),
        "Std CV F1": scores.std()
    })

cv_results_df = pd.DataFrame(cv_results)

cv_results_df = cv_results_df.sort_values(
    "Mean CV F1",
    ascending=False
)

print("Cross-Validation Results:")
display(cv_results_df)
```

```python
# ============================================================
# PHASE 9 — ADVANCED MODEL EVALUATION
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    roc_curve,
    precision_recall_curve
)

evaluation_results = []

for name, model in tuned_models.items():

    model.fit(X_train_smote, y_train_smote)

    y_pred = model.predict(X_test)

    if hasattr(model, "predict_proba"):
        y_prob = model.predict_proba(X_test)[:, 1]
    else:
        y_prob = model.decision_function(X_test)

    evaluation_results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred),
        "F1 Score": f1_score(y_test, y_pred),
        "ROC-AUC": roc_auc_score(y_test, y_prob)
    })

evaluation_df = pd.DataFrame(evaluation_results)

evaluation_df = evaluation_df.sort_values(
    "F1 Score",
    ascending=False
)

print("Advanced Evaluation Results:")
display(evaluation_df)
```

```python
# ============================================================
# ROC CURVES
# ============================================================

plt.figure(figsize=(8, 6))

for name, model in tuned_models.items():

    if hasattr(model, "predict_proba"):
        y_prob = model.predict_proba(X_test)[:, 1]
    else:
        y_prob = model.decision_function(X_test)

    fpr, tpr, _ = roc_curve(y_test, y_prob)

    plt.plot(
        fpr,
        tpr,
        label=name
    )

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curves")
plt.legend()

plt.show()
```

```python
# ============================================================
# PRECISION-RECALL CURVES
# ============================================================

plt.figure(figsize=(8, 6))

for name, model in tuned_models.items():

    if hasattr(model, "predict_proba"):
        y_prob = model.predict_proba(X_test)[:, 1]
    else:
        y_prob = model.decision_function(X_test)

    precision, recall, _ = precision_recall_curve(
        y_test,
        y_prob
    )

    plt.plot(
        recall,
        precision,
        label=name
    )

plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curves")
plt.legend()

plt.show()
```
